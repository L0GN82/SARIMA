```python
import numpy as np
import pandas as pd
from statsmodels.tsa.statespace.sarimax import SARIMAX
from statsmodels.tsa.stattools import adfuller
import matplotlib.pyplot as plt
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
import matplotlib.dates as mdates
from regression_utils import *
import os
```

```python
#Accessing the adjusted values from SARIMA Project
adjusted_df = pd.read_excel(r"C:\Users\LoganFrideley\OneDrive - East Texas Salt Water Disposal Company\Documents\SARIMA Project\Water Disposal History.xlsx", sheet_name='Sheet2')
adjusted = adjusted_df[['Date', 'Total Water Volume 1']]

mask = (adjusted['Date'] >= '2021-01') & (adjusted['Date'] <= '2025-12')
subset = adjusted[mask]

mask_valid = (adjusted['Date'] >= '2026-01') & (adjusted['Date'] <= '2026-07')
validation = adjusted[mask_valid]

os.chdir(r"/")
p, d, q = 0, 1, 0
P, D, Q, s = 0, 0, 0, 12
model = SARIMAX(subset['Total Water Volume 1'], order=(p, d, q), seasonal_order=(P, D, Q, s))
results = model.fit()
forecast_periods = 7
forecast = results.forecast(forecast_periods)
adjusted_stat = adj_R_squared(validation['Total Water Volume 1'], forecast, adjusted)

print(f'Adjust R-squared: {adjusted_stat}')
```
